```python
from mnist.model import LeNet5
import torch
import pandas as pd

m = LeNet5()
x = torch.randn(10, 1, 28, 28)
out = m(x)
print(out)
display(pd.DataFrame(out.detach().numpy(), columns=range(10)).round(3))

softmaxes = torch.softmax(out, dim=1)
print("SOFTMAXES")
print(softmaxes)

```
